```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hello my name is {self.name} and I am {self.age} years old")

    def change_name(self, new_name):
        self.old_name = self.name
        self.name = new_name
```

```python
person_1 = Person("raphael", 278)
person_1.introduce()
```

```python
person_2 = Person("elizabeth", 74)
person_2.introduce()
```

```python
class Employee(Person): # Employee inherits from Person
    def __init__(self, name, age, role):
        super().__init__(name, age)
        # get parent, use init from parent
        self._role = role

    def introduce(self):
        super().introduce() # -> function that prints
        print(f"and I work in {self._role}")

    def introduce_role(self):
        print(f"I work in {self._role}")
```

```python
employee = Employee('elizabeth', 70, 'data scientist')
employee.introduce()
print()
employee.introduce_role()
```

```python
employee.change_name('Raphael')
print(employee.name)
print(employee.old_name)
```

```python
employee._role = 'new role'
```

```python
class ShortList:
    def __init__(self, values):
        self.data = values
    
    def add_to_list(self, new_value):
        if len(self.data) > 3:
            raise Exception("List too long.")
        self.data.append(new_value)
```

# ToDo List

```python
class Task:
    def __init__(self, description, status):
        self.description = description
        self.status = status

    def complete(self):
        self.status = 'Completed'

class ToDoList:
    def __init__(self):
        self.task_list = []

    def add_task(self, description, status):
        new_task = Task(description, status)
        self.task_list.append(new_task)

    def validate_task_number(self, task_number):
        if not isinstance(task_number, int):
            raise Exception(f"Task number {task_number} must be an integer.")
        
        if task_number > len(self.task_list):
            raise Exception(f"Task number {task_number} is greater than number of tasks.")

    def complete_task(self, task_number):
        self.validate_task_number(task_number)
        task = self.task_list[task_number - 1]
        task.complete()

    def modify_task(self, task_number, new_description=None, new_status=None):
        self.validate_task_number(task_number)

        if not any([new_description, new_status]):
            raise Exception("You must update at least one information about the task!")

        task = self.task_list[task_number - 1]

        if new_description:
            task.description = new_description

        if new_status:
            task.status = new_status

    def display_tasks(self):
        for i, task in enumerate(self.task_list, 1):
            print(f"Task {i}: {task.description} | Status: {task.status}")
```

```python
todo_list = ToDoList()
todo_list.add_task(description="sortir les poubelles", status="Not done")
todo_list.add_task("étudier", "not done")
```

```python
todo_list.display_tasks()
```

```python
todo_list.complete_task(1)
todo_list.display_tasks()
```

```python
todo_list.modify_task(2)
todo_list.display_tasks()
```
