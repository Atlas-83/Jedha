### vérifier la date de commande la plus ancienne pour chaque kiosque pour s'assurer de la comparabilité

```python
# 1. Conversion des dates en datetime
commandes['date_commande'] = pd.to_datetime(commandes['date_commande'], errors='coerce')

# 2. Fonction de tri spécifique pour les IDs au format K01, K02, etc.
def trier_kiosque(id):
    """Extrait le numéro du kiosque pour le tri (ex: K01 -> 1)"""
    try:
        # Extraire le numéro et le convertir en entier
        return int(id[1:])  # On prend tout après le 'K' et on convertit en int
    except:
        return 999  # Valeur par défaut pour les IDs non conformes

# 3. Tri de la liste des kiosques
liste_kap_sorted = sorted(liste_kap, key=trier_kiosque)

# 4. Affichage des résultats triés
print("Dates de premières et dernières commandes par kiosque :")
print("=" * 80)

for kiosque_id in liste_kap_sorted:
    # Filtrer les commandes pour ce kiosque
    kiosque_commands = commandes[commandes['kiosque_id'] == kiosque_id]

    # Calculer les dates min/max
    first_date = kiosque_commands['date_commande'].min()
    last_date = kiosque_commands['date_commande'].max()

    # Calculer la durée d'activité
    if pd.notna(first_date) and pd.notna(last_date):
        duration = (last_date - first_date).days
        duration_str = f"{duration} jours"
    else:
        duration_str = "N/A"

    # Affichage formaté
    print(f"{kiosque_id}:")
    print(f"  Première commande: {first_date.strftime('%d/%m/%Y') if pd.notna(first_date) else 'N/A'}")
    print(f"  Dernière commande:  {last_date.strftime('%d/%m/%Y') if pd.notna(last_date) else 'N/A'}")
    print(f"  Durée d'activité:   {duration_str}")
    print("-" * 80)
```

```python
kiosque_filtre_id = "k04"
commandes_kiosque = commandes_periode_analyse[commandes_periode_analyse['kiosque_id'] == kiosque_filtre_id]

# 1. Calcul du revenu annuel (déjà fait dans votre code précédent)
revenu_annee_kiosque_filtre = commandes_kiosque.groupby("annee")["revenu"].sum().reset_index()
revenu_annee_kiosque_filtre = revenu_annee_kiosque_filtre.rename(columns={
    "annee": "Année",
    "revenu": "Revenu"
})

# 2. Nombre de commandes par année (format numérique)
nbcom_annee_kiosque_filtre = commandes_kiosque.groupby("annee")["num_commande"].nunique().reset_index()
nbcom_annee_kiosque_filtre = nbcom_annee_kiosque_filtre.rename(columns={
    "annee": "Année",
    "num_commande": "Nombre de commandes"
})

# 3. Calcul du panier moyen
panier_annee_kiosque_filtre = (revenu_annee_kiosque_filtre.set_index('Année')['Revenu'] /
                              nbcom_annee_kiosque_filtre.set_index('Année')['Nombre de commandes']).reset_index()
panier_annee_kiosque_filtre = panier_annee_kiosque_filtre.rename(columns={"index": "Année", 0: "Panier moyen"})

# Fonction pour formater les valeurs en euros
def format_euro(x):
    if pd.isna(x):
        return "0,00 €"
    return f"{x:,.2f} €".replace(",", " ").replace(".", ",")

# Fonction pour formater les nombres (sans décimales)
def format_nombre(x):
    if pd.isna(x):
        return "0"
    return f"{x:,.0f}".replace(",", " ")

# 4. Création des DataFrames finaux
# Nombre de commandes (format numérique)
nbcom_df = nbcom_annee_kiosque_filtre.copy()
nbcom_df['Nombre de commandes'] = nbcom_df['Nombre de commandes'].apply(format_nombre)

# Panier moyen (format monétaire)
panier_df = panier_annee_kiosque_filtre.copy()
panier_df['Panier moyen (€)'] = panier_df['Panier moyen'].apply(format_euro)

# Revenu annuel
revenu_df = revenu_annee_kiosque_filtre.copy()
revenu_df['Revenu (€)'] = revenu_df['Revenu'].apply(format_euro)
revenu_df = revenu_df[['Année', 'Revenu (€)']]  # Réorganisation des colonnes

# 5. Calcul des totaux
total_commandes = nbcom_annee_kiosque_filtre['Nombre de commandes'].sum()
total_panier = (revenu_annee_kiosque_filtre['Revenu'].sum() /
               nbcom_annee_kiosque_filtre['Nombre de commandes'].sum())


# 6. Ajout des lignes de total
# Pour le nombre de commandes
total_com_row = pd.DataFrame({
    'Année': ['TOTAL'],
    'Nombre de commandes': [format_nombre(total_commandes)]
})

# Pour le panier moyen
total_panier_row = pd.DataFrame({
    'Année': ['TOTAL'],
    'Panier moyen (€)': [format_euro(total_panier)]
})

# Pour le revenu annuel
total_revenu = revenu_annee_kiosque_filtre['Revenu'].sum()
total_row = pd.DataFrame({
    'Année': ['TOTAL'],
    'Revenu (€)': [format_euro(total_revenu)]
})

# 7. Concatenation avec les totaux
nbcom_df = pd.concat([
    nbcom_df[['Année', 'Nombre de commandes']],
    total_com_row
], ignore_index=True)

panier_df = pd.concat([
    panier_df[['Année', 'Panier moyen (€)']].rename(columns={'Panier moyen': 'Panier moyen (€)'}),
    total_panier_row
], ignore_index=True)


revenu_df = pd.concat([
    revenu_df, total_row], ignore_index=True)

# 8. Affichage des résultats
print(f"\n{'='*60}")
print(f"STATISTIQUES POUR LE KIOSQUE {kiosque_filtre_id.upper()}")
print(f"{'='*60}")

print("\nNombre annuel de commandes:")
display(nbcom_df)

print("\nPanier moyen annuel:")
display(panier_df)

print("\nRevenu annuel:")
display(revenu_df[['Année', 'Revenu (€)']])
```

```python
# 1. Fonctions de formatage
def format_euro(x):
    """Formate un nombre en euros avec le format français"""
    if pd.isna(x) or x == 0:
        return "0,00 €"
    return f"{x:,.2f} €".replace(",", " ").replace(".", ",")

def format_nombre(x):
    """Formate un nombre avec séparateurs de milliers"""
    if pd.isna(x):
        return "0"
    return f"{x:,.0f}".replace(",", " ")

# 2. Chiffre d'affaires annuel par source
print("\n" + "="*60)
print("CHIFFRE D'AFFAIRES ANNUEL PAR SOURCE")
print("="*60)

revenu_par_source = commandes_periode_analyse.groupby(["source", "annee"])["revenu"].sum().reset_index()
revenu_par_source["Revenu (€)"] = revenu_par_source["revenu"].apply(format_euro)
revenu_pivot = revenu_par_source.pivot(index="annee", columns="source", values="Revenu (€)")
display(revenu_pivot)

# 3. Nombre de commandes par source
print("\n" + "="*60)
print("NOMBRE DE COMMANDES PAR SOURCE")
print("="*60)

nb_com_par_source = commandes_periode_analyse.groupby(["source", "annee"])["num_commande"].nunique().reset_index()
nb_com_par_source["Nombre de commandes"] = nb_com_par_source["num_commande"].apply(format_nombre)
nb_com_pivot = nb_com_par_source.pivot(index="annee", columns="source", values="Nombre de commandes")
display(nb_com_pivot)

# 4. Panier moyen annuel par source
print("\n" + "="*60)
print("PANIER MOYEN ANNUEL PAR SOURCE")
print("="*60)

# Calcul du panier moyen
panier_moyen = (
    commandes_periode_analyse.groupby(["source", "annee"])["revenu"].sum() /
    commandes_periode_analyse.groupby(["source", "annee"])["num_commande"].nunique()
).reset_index(name="panier_moyen")

panier_moyen["Panier moyen (€)"] = panier_moyen["panier_moyen"].apply(format_euro)
panier_pivot = panier_moyen.pivot(index="annee", columns="source", values="Panier moyen (€)")
display(panier_pivot)

# 5. Version synthétique complète
print("\n" + "="*80)
print("SYNTHÈSE COMPLÈTE PAR SOURCE ET ANNÉE")
print("="*80)

# Création d'un DataFrame complet
synthese = commandes_periode_analyse.groupby(["source", "annee"]).agg({
    "revenu": "sum",
    "num_commande": "nunique"
}).reset_index()

synthese["panier_moyen"] = synthese["revenu"] / synthese["num_commande"]

# Formatage des colonnes
synthese["Revenu (€)"] = synthese["revenu"].apply(format_euro)
synthese["Nombre de commandes"] = synthese["num_commande"].apply(format_nombre)
synthese["Panier moyen (€)"] = synthese["panier_moyen"].apply(format_euro)

# Affichage par source
for source in synthese["source"].unique():
    print(f"\nSource: {source}")
    source_data = synthese[synthese["source"] == source]
    display(source_data[["annee", "Revenu (€)", "Nombre de commandes", "Panier moyen (€)"]])
```

# UPSELLING ET CROSS-SELLING

```python
commandes_periode_analyse['is_supplement'] = commandes_periode_analyse['id_article'].astype(str).str.startswith('SUP')

import numpy as np

# Colonne 'ca_supplement' : le revenu si c'est un SUP, 0 sinon
commandes_periode_analyse['ca_supplement'] = np.where(
    commandes_periode_analyse['is_supplement'] == True,
    commandes_periode_analyse['revenu'],
    0
)

sum_suppl = commandes_periode_analyse["ca_supplement"].sum()
print(sum_suppl)
```

```python
# analyse du upselling

part_suppl = ["source", "annee"]
suppl_ca = commandes_periode_analyse.groupby(part_suppl).agg(
    ca_global=("revenu", "sum"),
    ca_supplement=("ca_supplement", "sum"),
)
suppl_ca["part_suppl_ratio"] = (suppl_ca['ca_supplement'] / suppl_ca['ca_global'] * 100).round(2)

print("--- CA Global et CA Suppléments agrégés par Année")
print(suppl_ca)
```

```python
# analyse du cross_selling (Boissons et Desserts)
cond_boi = commandes_periode_analyse['id_article'].astype(str).str.startswith("BOI")
cond_des = commandes_periode_analyse['id_article'].astype(str).str.startswith("DES")
commandes_periode_analyse['is_cross'] = cond_boi | cond_des

# Colonne 'ca_cross' : le revenu si c'est un SUP, 0 sinon
commandes_periode_analyse['ca_cross'] = np.where(
    commandes_periode_analyse['is_cross'] == True,
    commandes_periode_analyse['revenu'],
    0
)

sum_cross = commandes_periode_analyse["ca_cross"].sum()
print(sum_cross)
```

```python
part_cross = ["source", "annee"]
cross_ca = commandes_periode_analyse.groupby(part_cross).agg(
    ca_cross_global=("revenu", "sum"),
    ca_cross=("ca_cross", "sum"),
)
cross_ca["part_cross_ratio"] = (cross_ca['ca_cross'] / cross_ca['ca_cross_global'] * 100).round(2)

print("--- CA Global et CA Suppléments agrégés par Année")
print(cross_ca)
```

### Conclusions : 
- un upselling stable entre 2023 et 2025
- la part du cross-selling dans le CA est moins importante en CAISSE que les autres sources
- un cross-selling stable ou en légère progression

```python
# 1. Définir les dates d'observation (du 01/04/2023 au 01/12/2025)
# 'MS' signifie 'Month Start' (le 1er du mois)
dates_observation = pd.date_range(start='2023-04-01', end='2025-12-01', freq='MS')

liste_resultats = []

print("Début du traitement par Kiosque...")

for date_obs in dates_observation:
    # 1. Fenêtre glissante [J-90, Date_Obs[
    date_debut = date_obs - pd.Timedelta(days=90)
    
    # 2. Filtrer les commandes de la période
    mask = (commandes['date_commande'] >= date_debut) & (commandes['date_commande'] < date_obs)
    df_window = commandes.loc[mask].copy()
    
    # 3. Calculer le nombre de commandes par Client
    fidelite = df_window.groupby('client_id').size().reset_index(name='nb_commandes_90j')
    
    # Filtrer uniquement les fidèles (>= 3 commandes)
    clients_fideles_ids = fidelite[fidelite['nb_commandes_90j'] >= 3]['client_id']
    
    # Si on a des clients fidèles ce mois-ci...
    if not clients_fideles_ids.empty:
        # On ne garde que les commandes de ces clients fidèles pour déterminer leur kiosque
        commandes_fideles = df_window[df_window['client_id'].isin(clients_fideles_ids)]
        
        # --- 4. DÉTERMINATION DU KIOSQUE DOMINANT ---
        # On trie par date (descendant) pour privilégier le plus récent en cas d'égalité de volume
        commandes_fideles = commandes_fideles.sort_values('date_commande', ascending=False)
        
        # On compte les commandes par Client ET par Kiosque
        kiosque_stats = commandes_fideles.groupby(['client_id', 'kiosque_id']).size().reset_index(name='count')
        
        # On trie pour avoir le kiosque le plus fréquent en premier pour chaque client
        kiosque_stats = kiosque_stats.sort_values(['client_id', 'count'], ascending=[True, False])
        
        # On dédoublonne pour ne garder que la 1ère ligne (le kiosque majoritaire) par client
        kiosque_dominant = kiosque_stats.drop_duplicates(subset=['client_id'], keep='first')[['client_id', 'kiosque_id']]
        
        # --- 5. Finalisation ---
        # On ajoute la date d'observation
        kiosque_dominant['date_observation'] = date_obs
        
        liste_resultats.append(kiosque_dominant)

# --- Consolidation ---
if liste_resultats:
    df_fidelite_kiosque = pd.concat(liste_resultats, ignore_index=True)
    
    print("\n--- Aperçu (Client affecté à son kiosque du moment) ---")
    print(df_fidelite_kiosque.head())
```

```python
   # --- Analyse Croisée : Évolution par Kiosque ---
evolution_par_kiosque = df_fidelite_kiosque.groupby(['date_observation', 'kiosque_id']).size().unstack(fill_value=0)
    
print("\n--- Évolution des Clients Fidèles par Kiosque ---")
print(evolution_par_kiosque)

evolution_par_kiosque.to_excel("export_clients_fideles.xlsx")
```

```python
#Script pour calcul du taux d'attrition

# --- Hypothèses ---
# 1. 'df_fidelite_kiosque' est le résultat du script précédent (colonnes : client_id, kiosque_id, date_observation)
# 2. 'commandes' est votre base de commandes (avec date_commande, client_id)

liste_attrition = []

# On récupère toutes les dates d'observation (les 1er du mois) présentes dans votre étude
dates_uniques = sorted(df_fidelite_kiosque['date_observation'].unique())

print("Calcul de l'attrition en cours...")

for date_debut_mois in dates_uniques:
    # 1. Définir la période de "surveillance" (le mois courant complet)
    # Ex: Si date_debut_mois = 01/04/2023, on regarde jusqu'au 01/05/2023 exclus
    date_fin_mois = date_debut_mois + pd.DateOffset(months=1)
    
    # 2. Isoler les clients fidèles identifiés à ce début de mois
    # On fait une copie pour éviter les avertissements (SettingWithCopyWarning)
    fideles_du_mois = df_fidelite_kiosque[
        df_fidelite_kiosque['date_observation'] == date_debut_mois
    ].copy()
    
    if fideles_du_mois.empty:
        continue

    # 3. Identifier QUI a commandé durant ce mois parmi TOUS les clients
    commandes_du_mois = commandes[
        (commandes['date_commande'] >= date_debut_mois) & 
        (commandes['date_commande'] < date_fin_mois)
    ]
    clients_actifs_ids = set(commandes_du_mois['client_id'].unique())
    
    # 4. Taguer chaque client fidèle : a-t-il commandé ou non ?
    # La colonne 'is_perdu' est Vraie si le client N'EST PAS dans les actifs
    fideles_du_mois['is_perdu'] = ~fideles_du_mois['client_id'].isin(clients_actifs_ids)
    
    # 5. Agréger les résultats par Kiosque
    synthese_mois = fideles_du_mois.groupby(['date_observation', 'kiosque_id']).agg(
        nb_fideles_debut=('client_id', 'count'),
        nb_perdus_fin_mois=('is_perdu', 'sum') # La somme des True donne le nombre de perdus
    ).reset_index()
    
    # Calcul du taux
    synthese_mois['taux_attrition'] = (synthese_mois['nb_perdus_fin_mois'] / synthese_mois['nb_fideles_debut']) * 100
    
    liste_attrition.append(synthese_mois)

# --- Consolidation ---
df_attrition = pd.concat(liste_attrition, ignore_index=True)

# Mise en forme pour l'affichage
pd.options.display.float_format = '{:.2f}%'.format # Format pourcentage
print("\n--- Taux d'Attrition Mensuel des Clients Fidèles par Kiosque ---")
print(df_attrition.head(10))

# --- Optionnel : Vue globale (tous kiosques confondus) ---
df_attrition_global = df_attrition.groupby('date_observation').agg({
    'nb_fideles_debut': 'sum',
    'nb_perdus_fin_mois': 'sum'
}).reset_index()
df_attrition_global['taux_attrition_global'] = (df_attrition_global['nb_perdus_fin_mois'] / df_attrition_global['nb_fideles_debut']) * 100

print("\n--- Évolution Globale du Taux d'Attrition ---")
print(df_attrition_global.head())

df_attrition.to_excel("export_attrition.xlsx")
```
